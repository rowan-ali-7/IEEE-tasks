```
import numpy as np
grades = [
    [85,78,92,88],
    [70,76,80,65],
    [90,88,94,91],
    [60,65,58,62],
    [100,95,98,97]
]
Grades = np.array(grades)
print("Shape : ", Grades.shape)
student_mean=Grades.mean(axis=1)
print("Mean for each student : ",student_mean)
print("Mean for each subject : " , Grades.mean(axis=0))
print("Students whose average grade is greater than 85 :\n " , Grades[student_mean>85])
print("After bonus : \n" ,Grades+5)
min=Grades.min(axis=0)
max=Grades.max(axis=0)
normalized_grades = (Grades - min)/(max - min)
print("After normalization : \n" ,normalized_grades)
Flatten_grades=Grades.reshape(-1)
print("Flatten : \n" , Flatten_grades)
```

```
x = [50,60,80,100,120]
y = [150,180,240,300,330]
house_size = np.array(x)
house_size = house_size[: , np.newaxis]
price = np.array(y)
one = np.ones((5,1))
bais_x = np.concatenate((one,house_size),axis=1)
theta = (np.linalg.inv(((bais_x.T)@bais_x))) @ ((bais_x.T)@price)
new_pred = np.array([1,90])
pred = new_pred@theta
print(pred)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1/(1+np.e**(-z))
x = np.linspace(-7, 7, 200)
y = sigmoid(x)
```

```python
plt.figure(figsize=(8,5))
plt.plot(x, y, label="Sigmoid Function",color="#f3b8d1", linewidth=2)
plt.title("Sigmoid Function")
plt.xlabel("x")
plt.ylabel("Sigmoid")
plt.grid(True)
plt.show()
```

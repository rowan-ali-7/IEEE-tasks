```python
def matsum (arr1 , arr2):
    arr3 = [[0 for i in range(len(arr1[0]))]for i in range(len(arr1))]
    for i in range(len(arr1)):
        for j in range(len(arr1[0])):
            arr3[i][j] = arr1[i][j] + arr2[i][j]
    return arr3
arr1 = [
    [5,2,6],
    [7,4,5],
    [9,1,4]
]

arr2 = [
    [5,7,6],
    [7,4,7],
    [7,1,4]
]

print(matsum(arr1 , arr2))
```

```python
def matsub (arr1 , arr2):
    arr3 = [[0 for i in range(len(arr1[0]))]for i in range(len(arr1))]
    for i in range(len(arr1)):
        for j in range(len(arr1[0])):
            arr3[i][j] = arr1[i][j] - arr2[i][j]
    return arr3
arr1 = [
    [5,2,6],
    [7,4,5],
    [9,1,4]
]

arr2 = [
    [5,7,6],
    [7,4,7],
    [7,1,4]
]

print(matsub(arr1 , arr2))
```

```python
def  scalarsum(scalar , arr) :
    Sarr = [[0 for i in range(len(arr[0]))]for j in range(len(arr))]
    for i in range(len(arr)):
        for j in range(len(arr[0])):
            Sarr[i][j] = scalar + arr[i][j]
    return Sarr
 
arr = [
    [5,2,6],
    [7,4,5],
    [9,1,4]
]
print(scalarsum(7, arr))
```

```python
def  scalarsub(scalar , arr) :
    Sarr = [[0 for i in range(len(arr[0]))]for j in range(len(arr))]
    for i in range(len(arr)):
        for j in range(len(arr[0])):
            Sarr[i][j] = arr[i][j] - scalar 
    return Sarr
 
arr = [
    [5,2,6],
    [7,4,5],
    [9,1,4]
]
print(scalarsub(7 , arr))
```

```python
def matmul(arr1 , arr2):
    arr3 = [[0 for i in range(len(arr2[0]))]for i in range(len(arr1))]
    for i in range(len(arr1)):
        for j in range(len(arr2[0])):
            sum = 0
            for n in range(len(arr1[0])):
                sum += arr1[i][n] * arr2[n][j]
            arr3[i][j] = sum
    return arr3

arr1 = [
    [5,2,6],
    [7,4,5],
    [9,1,4]
]

arr2 = [
    [5,7,6],
    [7,4,7],
    [7,1,4]
]

print(matmul(arr1 , arr2))
```
